```python
from pathlib import Path
import sys

import anndata as ad
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.patches import FancyBboxPatch


# ============================================================
# 1. Locate repository
# ============================================================

def find_repo_root(start=Path.cwd()):
    start = Path(start).resolve()

    for p in [start, *start.parents]:
        if (p / "src" / "scrna_benchmark").is_dir():
            return p

    raise RuntimeError(
        "Run this notebook from somewhere inside the repository."
    )


REPO_ROOT = find_repo_root()

if str(REPO_ROOT / "src") not in sys.path:
    sys.path.insert(
        0,
        str(REPO_ROOT / "src"),
    )


# ============================================================
# 2. Paths
# ============================================================

ADATA_PATH = (
    REPO_ROOT
    / "data/lung/lung_benchmark_ready.h5ad"
)

PRIMARY_DIR = (
    REPO_ROOT
    / "results/lung_no_batch_L4"
)

BATCH_DIR = (
    REPO_ROOT
    / "results/lung_with_batch_L4"
)

RANDOM_PATH = (
    PRIMARY_DIR
    / "random_split/random_split_repeated_metrics.csv"
)

DONOR_PATH = (
    PRIMARY_DIR
    / "donor_cv/metrics.csv"
)

BATCH_RANDOM_PATH = (
    BATCH_DIR
    / "random_split/random_split_repeated_metrics.csv"
)

BATCH_DONOR_PATH = (
    BATCH_DIR
    / "donor_cv/metrics.csv"
)

GLMM_PATH = (
    PRIMARY_DIR
    / "mixed_models/correctness_glmm_predicted_probabilities.csv"
)

ABLATION_PATH = (
    PRIMARY_DIR
    / "donor_ablation/donor_ablation_summary.csv"
)

FIG_DIR = (
    REPO_ROOT
    / "manuscript_figures"
)


# ============================================================
# 3. Validate inputs
# ============================================================

required = [
    ADATA_PATH,
    RANDOM_PATH,
    DONOR_PATH,
    BATCH_RANDOM_PATH,
    BATCH_DONOR_PATH,
    GLMM_PATH,
    ABLATION_PATH,
]

missing = [
    p for p in required
    if not p.exists()
]

if missing:
    raise FileNotFoundError(
        "Missing required Figure 3 inputs:\n"
        + "\n".join(str(p) for p in missing)
    )


# ============================================================
# 4. Cohort information
# ============================================================

adata = ad.read_h5ad(
    ADATA_PATH,
    backed="r",
)

n_cells = adata.n_obs

n_donors = (
    adata.obs["donor_id"]
    .astype(str)
    .nunique()
)

n_cell_types = (
    adata.obs["cell_type"]
    .astype(str)
    .nunique()
)

n_source_datasets = (
    adata.obs["batch"]
    .astype(str)
    .nunique()
)

# Optional: useful for checking source-dataset composition.
source_summary = (
    pd.DataFrame({
        "batch": adata.obs["batch"].astype(str),
        "donor_id": adata.obs["donor_id"].astype(str),
    })
    .groupby(
        "batch",
        observed=True,
    )
    .agg(
        n_cells=("batch", "size"),
        n_donors=("donor_id", "nunique"),
    )
)

if (
    hasattr(adata, "file")
    and adata.file is not None
):
    adata.file.close()


# ============================================================
# 5. Load numerical results
# ============================================================

random_df = pd.read_csv(
    RANDOM_PATH
)

donor_df = pd.read_csv(
    DONOR_PATH
)

batch_random_df = pd.read_csv(
    BATCH_RANDOM_PATH
)

batch_donor_df = pd.read_csv(
    BATCH_DONOR_PATH
)

glmm_df = pd.read_csv(
    GLMM_PATH
)

ablation_df = pd.read_csv(
    ABLATION_PATH
)


# ============================================================
# 6. Representation order
# ============================================================

REP_ORDER = [
    "hvg",
    "pca",
    "harmony",
    "scvi",
]

REP_LABELS = {
    "hvg": "HVG",
    "pca": "PCA",
    "harmony": "Harmony",
    "scvi": "scVI",
}


# ============================================================
# 7. Panel B summaries
#    Random cell-level split vs donor-held-out CV
# ============================================================

random_summary = (
    random_df
    .groupby("representation")["macro_f1"]
    .agg(["mean", "std"])
    .reindex(REP_ORDER)
)

donor_summary = (
    donor_df
    .set_index("representation")
    .reindex(REP_ORDER)
)


# ============================================================
# 8. Panel C summaries
#    Source-dataset covariate sensitivity
# ============================================================

batch_random_summary = (
    batch_random_df
    .groupby("representation")["macro_f1"]
    .mean()
    .reindex(REP_ORDER)
)

batch_donor_summary = (
    batch_donor_df
    .set_index("representation")["macro_f1_mean"]
    .reindex(REP_ORDER)
)

random_delta = (
    batch_random_summary
    - random_summary["mean"]
)

donor_delta = (
    batch_donor_summary
    - donor_summary["macro_f1_mean"]
)


# ============================================================
# 9. Panel D summaries
#    Donor random-intercept GLMM
# ============================================================

glmm_plot = glmm_df.copy()

glmm_plot["representation"] = (
    glmm_plot["representation"]
    .astype(str)
)

glmm_plot["scheme"] = (
    glmm_plot["scheme"]
    .astype(str)
)

scheme_rename = {
    "random": "Random split",
    "random_split": "Random split",

    "donor_held_out": "Donor-held-out",
    "donor-held-out": "Donor-held-out",
    "donor_cv": "Donor-held-out",
}

glmm_plot["scheme_clean"] = (
    glmm_plot["scheme"]
    .map(scheme_rename)
    .fillna(glmm_plot["scheme"])
)

glmm_wide = (
    glmm_plot
    .pivot(
        index="representation",
        columns="scheme_clean",
        values="pred_prob_correct",
    )
    .reindex(REP_ORDER)
)


# ============================================================
# 10. Styling
# ============================================================

plt.rcParams.update({
    "font.size": 10,
    "axes.titlesize": 11.5,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 8.5,
})

BLUE = "#1f77b4"
ORANGE = "#ff7f0e"
GRAY = "#666666"


def clean_axis(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


def panel_letter(
    ax,
    letter,
    x=-0.08,
    y=1.03,
):
    ax.text(
        x,
        y,
        letter,
        transform=ax.transAxes,
        fontsize=15,
        fontweight="bold",
        va="bottom",
        ha="left",
        clip_on=False,
    )


# ============================================================
# 11. Figure layout
#
# Same visual grammar as Figure 2:
#
#    A |                 B
#   --------------------------------
#          C       |   D   |   E
# ============================================================

fig = plt.figure(
    figsize=(17, 9.2),
    layout="constrained",
)

outer = fig.add_gridspec(
    2,
    14,
    height_ratios=[
        0.95,
        1.05,
    ],
)


# Top row
axA = fig.add_subplot(
    outer[0, 0:4]
)

axB = fig.add_subplot(
    outer[0, 4:14]
)


# Bottom row
axC = fig.add_subplot(
    outer[1, 0:6]
)

axD = fig.add_subplot(
    outer[1, 6:10]
)

axE = fig.add_subplot(
    outer[1, 10:14]
)


# ============================================================
# A. Lung cohort overview
# ============================================================

panel_letter(
    axA,
    "A",
    x=-0.02,
    y=1.00,
)

axA.axis("off")

axA.set_xlim(
    0,
    1,
)

axA.set_ylim(
    0,
    1,
)


axA.text(
    0.05,
    0.90,
    "Lung cohort",
    fontsize=13,
    fontweight="bold",
    va="top",
)


stats = [
    (
        f"{n_cells:,}",
        "cells",
    ),
    (
        f"{n_donors}",
        "donors",
    ),
    (
        f"{n_cell_types}",
        "cell types",
    ),
]


for x0, (big, small) in zip(
    [0.05, 0.35, 0.65],
    stats,
):

    box = FancyBboxPatch(
        (
            x0,
            0.65,
        ),
        0.25,
        0.16,
        boxstyle=(
            "round,"
            "pad=0.012,"
            "rounding_size=0.02"
        ),
        linewidth=1.0,
        facecolor="white",
    )

    axA.add_patch(
        box
    )

    axA.text(
        x0 + 0.125,
        0.735,
        big,
        ha="center",
        va="center",
        fontsize=13,
        fontweight="bold",
    )

    axA.text(
        x0 + 0.125,
        0.680,
        small,
        ha="center",
        va="center",
        fontsize=9,
    )


# Source-dataset information replaces the PBMC site section.
axA.text(
    0.05,
    0.55,
    "Source datasets",
    fontweight="bold",
)

axA.text(
    0.05,
    0.46,
    f"{n_source_datasets} datasets",
    fontsize=10,
)

axA.text(
    0.05,
    0.28,
    "Representations",
    fontweight="bold",
)

axA.text(
    0.05,
    0.18,
    "HVG   ·   PCA   ·   Harmony   ·   scVI",
    fontsize=10,
)


# ============================================================
# B. Evaluation strategy
# ============================================================

panel_letter(
    axB,
    "B",
    x=-0.06,
    y=1.00,
)


x = np.arange(
    len(REP_ORDER)
)

bar_width = 0.36


axB.bar(
    x - bar_width / 2,
    random_summary[
        "mean"
    ].to_numpy(),
    width=bar_width,
    yerr=random_summary[
        "std"
    ].to_numpy(),
    capsize=4,
    color=BLUE,
    edgecolor="black",
    linewidth=0.7,
    label="Random cell-level split",
)


axB.bar(
    x + bar_width / 2,
    donor_summary[
        "macro_f1_mean"
    ].to_numpy(),
    width=bar_width,
    yerr=donor_summary[
        "macro_f1_std"
    ].to_numpy(),
    capsize=4,
    color=ORANGE,
    edgecolor="black",
    linewidth=0.7,
    label="Donor-held-out CV",
)


axB.set_xticks(
    x,
    [
        REP_LABELS[r]
        for r in REP_ORDER
    ],
)


axB.set_ylabel(
    "Macro F1"
)

axB.set_xlabel(
    "Representation"
)

axB.set_title(
    "Evaluation strategy",
    pad=8,
)


# Dynamic but tight enough to preserve visible differences.
b_min = min(
    random_summary["mean"].min()
    - random_summary["std"].max(),
    (
        donor_summary["macro_f1_mean"]
        - donor_summary["macro_f1_std"]
    ).min(),
)

b_max = max(
    random_summary["mean"].max()
    + random_summary["std"].max(),
    (
        donor_summary["macro_f1_mean"]
        + donor_summary["macro_f1_std"]
    ).max(),
)

axB.set_ylim(
    max(0, b_min - 0.03),
    min(1, b_max + 0.03),
)


axB.legend(
    frameon=False,
    loc="upper right",
)

clean_axis(
    axB
)


# ============================================================
# C. Source-dataset covariate sensitivity
#
# Same single-axis structure as Figure 2.
# Left group  = random cell-level split
# Right group = donor-held-out CV
# ============================================================

panel_letter(
    axC,
    "C",
    x=-0.08,
    y=1.00,
)


# Random-split positions
x_random = np.arange(
    len(REP_ORDER)
)

# Intentional gap before donor-CV group
x_donor = (
    np.arange(
        len(REP_ORDER)
    )
    + 5.2
)

dx = 0.08


c_ymin = (
    min(
        random_summary["mean"].min(),
        donor_summary[
            "macro_f1_mean"
        ].min(),
        batch_random_summary.min(),
        batch_donor_summary.min(),
    )
    - 0.025
)

c_ymax = (
    max(
        random_summary["mean"].max(),
        donor_summary[
            "macro_f1_mean"
        ].max(),
        batch_random_summary.max(),
        batch_donor_summary.max(),
    )
    + 0.035
)


def add_covariate_group(
    ax,
    positions,
    base_values,
    cov_values,
    deltas,
):

    for xpos, rep in zip(
        positions,
        REP_ORDER,
    ):

        y_base = base_values.loc[
            rep
        ]

        y_cov = cov_values.loc[
            rep
        ]


        # Paired connector
        ax.plot(
            [
                xpos - dx,
                xpos + dx,
            ],
            [
                y_base,
                y_cov,
            ],
            color=GRAY,
            linewidth=1.3,
            zorder=1,
        )


        # No source-dataset covariate
        ax.scatter(
            xpos - dx,
            y_base,
            s=42,
            color=BLUE,
            zorder=3,
        )


        # Source-dataset covariate included
        ax.scatter(
            xpos + dx,
            y_cov,
            s=42,
            color=ORANGE,
            zorder=3,
        )


        delta = deltas.loc[
            rep
        ]

        ax.text(
            xpos,
            max(
                y_base,
                y_cov,
            ) + 0.007,
            f"Δ {delta:+.3f}",
            ha="center",
            va="bottom",
            fontsize=7.5,
        )


add_covariate_group(
    axC,
    x_random,
    random_summary[
        "mean"
    ],
    batch_random_summary,
    random_delta,
)


add_covariate_group(
    axC,
    x_donor,
    donor_summary[
        "macro_f1_mean"
    ],
    batch_donor_summary,
    donor_delta,
)


# Divider between evaluation schemes
axC.axvline(
    4.1,
    color="0.85",
    linewidth=1.0,
)


all_positions = np.concatenate([
    x_random,
    x_donor,
])


all_labels = (
    [
        REP_LABELS[r]
        for r in REP_ORDER
    ]
    * 2
)


axC.set_xticks(
    all_positions,
    all_labels,
    rotation=18,
)


axC.set_ylim(
    c_ymin,
    c_ymax,
)


axC.set_ylabel(
    "Macro F1"
)


axC.set_title(
    "Source-dataset covariate sensitivity",
    pad=24,
)


# Group headers
axC.text(
    np.mean(x_random),
    c_ymax - 0.005,
    "Random cell-level split",
    ha="center",
    va="top",
    fontsize=10.5,
)

axC.text(
    np.mean(x_donor),
    c_ymax - 0.005,
    "Donor-held-out CV",
    ha="center",
    va="top",
    fontsize=10.5,
)


handles_C = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        color=BLUE,
        markersize=6,
        label="No source covariate",
    ),
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        color=ORANGE,
        markersize=6,
        label="Source covariate included",
    ),
]


axC.legend(
    handles=handles_C,
    frameon=False,
    loc="lower left",
    ncol=2,
    fontsize=8,
)


clean_axis(
    axC
)


# ============================================================
# D. Donor random-intercept GLMM
# ============================================================

panel_letter(
    axD,
    "D",
    x=-0.08,
    y=1.00,
)


ypos = np.arange(
    len(REP_ORDER)
)


for i, rep in enumerate(
    REP_ORDER
):

    donor_val = glmm_wide.loc[
        rep,
        "Donor-held-out",
    ]

    random_val = glmm_wide.loc[
        rep,
        "Random split",
    ]


    # Connector emphasizes the within-representation
    # difference between evaluation strategies.
    axD.plot(
        [
            donor_val,
            random_val,
        ],
        [
            i,
            i,
        ],
        color=GRAY,
        linewidth=1.8,
        zorder=1,
    )


    # Random cell-level split
    axD.scatter(
        random_val,
        i,
        s=48,
        color=BLUE,
        zorder=3,
    )


    # Donor-held-out CV
    axD.scatter(
        donor_val,
        i,
        s=48,
        color=ORANGE,
        zorder=3,
    )


    delta = (
        random_val
        - donor_val
    )


    axD.text(
        random_val + 0.0015,
        i,
        f"+{delta:.3f}",
        va="center",
        ha="left",
        fontsize=8,
    )


axD.set_yticks(
    ypos,
    [
        REP_LABELS[r]
        for r in REP_ORDER
    ],
)

axD.invert_yaxis()


axD.set_xlabel(
    "GLMM predicted P(correct)"
)


# Slightly more title padding to create room
# for the legend below it.
axD.set_title(
    "Donor random-intercept GLMM",
    pad=28,
)


glmm_min = (
    glmm_wide
    .min()
    .min()
)

glmm_max = (
    glmm_wide
    .max()
    .max()
)


axD.set_xlim(
    glmm_min - 0.01,
    glmm_max + 0.015,
)


handles_D = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        color=BLUE,
        markersize=6,
        label="Random split",
    ),
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        color=ORANGE,
        markersize=6,
        label="Donor-held-out",
    ),
]


# Move legend upward, above the actual data rows.
axD.legend(
    handles=handles_D,
    frameon=False,
    loc="lower center",
    bbox_to_anchor=(0.5, 1.005),
    ncol=2,
    fontsize=8,
    handletextpad=0.5,
    columnspacing=1.2,
)


clean_axis(
    axD
)

# ============================================================
# E. Donor ablation
# ============================================================

panel_letter(
    axE,
    "E",
    x=-0.08,
    y=1.00,
)


for rep in REP_ORDER:

    sub = (
        ablation_df[
            ablation_df[
                "representation"
            ].eq(rep)
        ]
        .sort_values(
            "k_train_donors"
        )
    )


    axE.errorbar(
        sub[
            "k_train_donors"
        ],
        sub[
            "macro_f1_mean"
        ],
        yerr=sub[
            "macro_f1_std"
        ],
        marker="o",
        linewidth=1.6,
        capsize=3,
        label=REP_LABELS[rep],
    )


axE.set_xlabel(
    "Number of training donors"
)

axE.set_ylabel(
    "Macro F1"
)

axE.set_title(
    "Donor ablation",
    pad=8,
)


axE.set_xticks(
    sorted(
        ablation_df[
            "k_train_donors"
        ].unique()
    )
)


# Dynamic range because lung performance is higher than PBMC.
e_lower = (
    (
        ablation_df["macro_f1_mean"]
        - ablation_df["macro_f1_std"]
    )
    .min()
    - 0.03
)

e_upper = (
    (
        ablation_df["macro_f1_mean"]
        + ablation_df["macro_f1_std"]
    )
    .max()
    + 0.02
)

axE.set_ylim(
    max(0, e_lower),
    min(1, e_upper),
)


axE.legend(
    frameon=False,
    ncol=2,
    fontsize=8,
    title="Representation",
    title_fontsize=8,
    loc="lower right",
)


clean_axis(
    axE
)


# ============================================================
# 12. Display only for now
# ============================================================

# out_png = FIG_DIR / "figure3_lung.png"
# out_pdf = FIG_DIR / "figure3_lung.pdf"

# fig.savefig(
#     out_png,
#     dpi=300,
#     bbox_inches="tight",
# )

# fig.savefig(
#     out_pdf,
#     bbox_inches="tight",
# )

plt.show()
```
